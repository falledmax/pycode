# 总结梯度下降与正规方程法

* 对于梯度下降：其实权重和偏置电脑自身也不知道是多少，随机值，所以需要损失函数的求导得以验证距离最佳参数的距离，求导是判断损失函数的变化率，通过损失函数的增减性，来辅助寻炸wb
* 这就是与正规方程的差异，正规方程直接斜率=0，找到最低点，得到wb，梯度下降则是先瞎蒙wb，根据损失函数的变化调整wb,逐渐逼近最低点
* 梯度下降公式：下一个梯度=当前梯度-学习率*损失函数的导数 减号从何而来？用一个例子解释
    * 损失函数假设b＝0，只有w一个参数
它是绝对的开口向上的抛物线
它求导就是w的变化率 假设最棒的w是5，蒙的是2，新的w等于2-0.001乘以2乘负数，因为2在5左边单调下降所以导数是负的，
    * 样本的点无法改变 改变的只有w和b

梯度下降的四种算法：
* 全梯度下降算法（FGD）：每次调整wb的时候，会计算所有样本对wb的影响 假设有100个就用100个
* 随机梯度下降算法（SGD）：每次调整wb随机抽取一个样本，计算对wb的影响 假设有100个，从里面抽1个
* 小批量梯度下降算法（）：每次用min<x<max个样本，比最小大，比最大小，10个样本用9个照样是小梯度下降算法（用的最多）
* 随机平均梯度下降算法（SAG）：如果有N个样本，内存中就会有N个格子存储对应样本的梯度值（初始值设为0），每次调整wb的时候，在这N个样本里抽，抽到哪个样本对应的梯度值就更新哪个，取平均值（一般用‘动态分母’，有几个格子占着就除以几，格子沾满了，除以总数），格子沾满，不代表与FGD相同，就算沾满，SAG也是每次调整的时候抽一个，用来更新表。FGD是每次更新用全部数据，本质不同。
