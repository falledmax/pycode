# 归一化

适用于小数据集 且范围明确 且异常值几乎没有 例如 20岁 21岁 24岁 106岁 中的106 没有他


归一化：为了防止量纲单位问题，导致方差值相差过大 影响模型最终结果所以要把 各值映射到[0，1]之间
弊端是：容易受到最大值和最小值的影响 所以一般用于处理 小数据集

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler  # 归一化对象
x_train=[[90,2,10,40],[60,4,15,45],[75,3,13,46]]

#创建归一化对象
scaler=MinMaxScaler(feature_range=(3,5))
#feature_range=(3,5) 设置范围是3到5 可以设置生成范围 默认为0，1 如果就是这个区间可以忽略不写
#改变的是第二个公式的max和min 设置成了3和5


#对数据进行归一化操作
x_train_new=scaler.fit_transform(x_train)
print(x_train_new)
```

# 标准化

标准化处理 归一化：为了防止量纲单位问题，导致方差值相差过大影响模型最终结果
所以通过公式把 各列的值 映射到 均值为0 标准差为1的 正态分布序列
多应用于大数据的处理

```python
from sklearn.preprocessing import StandardScaler #标准化对象

#准备数据
x_train=[[90,2,10,40],[60,4,15,45],[75,3,13,46]]

#创建归一化对象
scaler=StandardScaler()

#对数据进行标准化操作
x_train_new=scaler.fit_transform(x_train)
print(x_train_new)
#方差 该列每个值与该列的平均值的差 的平方和的 平均值
#标准差 方差开平方根
```
