# 梯度下降

# 三种梯度下降方式

梯度下降（Gradient Descent）的核心：**计算梯度，并沿着梯度的反方向更新模型参数，使损失函数 $L$ 逐渐减小。** 参数更新公式：$W_{new}=W_{old}-\eta\frac{\partial L}{\partial W}$，其中 $W$ 是模型参数，$\eta$ 是学习率，$\frac{\partial L}{\partial W}$ 是梯度。三种梯度下降方式最核心的区别：**一次使用多少个样本计算梯度。**

| 方法 | Batch Size | 一次使用的数据 | $N=10000$ 时每个 Epoch 更新次数 | 特点 |
|---|---:|---|---:|---|
| BGD（批量梯度下降） | $N$ | 全部训练集 | $1$ | 梯度稳定，但计算量大 |
| SGD（随机梯度下降） | $1$ | 1 个样本 | $10000$ | 更新频繁，但梯度波动大 |
| Mini-Batch（小批量梯度下降） | $B$ | 一小批样本 | $\lceil N/B\rceil$ | 兼顾速度和稳定性，深度学习最常用 |

### 1. BGD：批量梯度下降

假设训练集有 **10000 个样本**，BGD 一次把全部 10000 个样本用于计算 Loss 和梯度，然后更新一次参数。

`10000 个样本 → 计算梯度 → 更新参数`

因此 `Batch Size = 10000`，一个 Epoch 只更新 **1 次**。例如训练 10 个 Epoch，大约更新 10 次参数。BGD 的优点是梯度比较稳定；缺点是每次计算量很大，数据集较大时速度较慢、内存占用较高。

### 2. SGD：随机梯度下降

还是 10000 个样本，SGD 每次只使用 **1 个样本**计算梯度，然后立即更新参数。

`样本1 → 计算梯度 → 更新参数`
`样本2 → 计算梯度 → 更新参数`
`样本3 → 计算梯度 → 更新参数`
`...`
`样本10000 → 计算梯度 → 更新参数`

因此 `Batch Size = 1`，一个 Epoch 大约更新 **10000 次参数**。由于每次只根据一个样本计算梯度，单个样本可能对梯度产生较大影响，所以 SGD 的梯度波动通常比较大。

### 3. Mini-Batch：小批量梯度下降

Mini-Batch 介于 BGD 和 SGD 之间。例如训练集有 10000 个样本，设置 `Batch Size = 32`，每次取 32 个样本计算梯度并更新参数。

`32 个样本 → 计算梯度 → 更新参数`
`32 个样本 → 计算梯度 → 更新参数`
`32 个样本 → 计算梯度 → 更新参数`
`...`

一个 Epoch 的 Batch 数量为 $\lceil10000/32\rceil=313$，因此一个 Epoch 大约更新 **313 次参数**。如果最后剩余样本不足 32 个，也可以组成最后一个 Batch，所以 Batch 数量通常为 $\lceil N/B\rceil$。

### 4. 用一个例子理解

假设你有 **10000 道题**，需要根据做题情况不断调整自己的学习方法：

- **BGD**：10000 道题全部做完 → 总结一次 → 调整学习方法
- **SGD**：做 1 道题 → 总结一次 → 调整 → 再做 1 道 → 再调整
- **Mini-Batch**：每做 32 道题 → 总结一次 → 调整 → 再做 32 道 → 再调整

所以三者本质上的区别就是 **Batch Size 不同**。

### 5. PyTorch 中的 Mini-Batch

实际深度学习中最常使用 Mini-Batch，例如 `DataLoader(dataset, batch_size=32)`。

其中 `batch_size=32` 表示：**每次取 32 个样本 → 前向传播 → 计算 Loss → 反向传播计算梯度 → 更新模型参数。**

### 6. 一句话记忆

**BGD：一次用全部；SGD：一次用一个；Mini-Batch：一次用一小批。**

可以简单理解为：**BGD 一次吃完整桌，SGD 一次吃一口，Mini-Batch 一次吃一小碗。**
