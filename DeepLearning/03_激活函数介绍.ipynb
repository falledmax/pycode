# Sigmoid 激活函数

## 1. Sigmoid 是什么？

Sigmoid 是神经网络中一种常见的激活函数，公式是：

$$
\sigma(x)=\frac{1}{1+e^{-x}}
$$

它最简单的作用就是：**把任意大小的数压缩到 0～1 之间**。例如 $x=-10$ 时，$\sigma(x)\approx0$；$x=0$ 时，$\sigma(x)=0.5$；$x=10$ 时，$\sigma(x)\approx1$。所以 Sigmoid 的输出范围是 $(0,1)$。

## 2. 为什么需要 Sigmoid？

神经网络中的神经元首先会进行加权求和：

$$
z=w_1x_1+w_2x_2+\cdots+w_nx_n+b
$$

如果直接把 $z$ 传给下一层，那么多层线性计算最终仍然可以合并成一个线性函数，网络就无法很好地学习复杂的非线性关系。因此需要加入激活函数，让神经网络具有**非线性表达能力**。Sigmoid 就是一种非线性激活函数。

## 3. 为什么 x 越大，Sigmoid 输出越大？

因为：

$$
\sigma(x)=\frac{1}{1+e^{-x}}
$$

当 $x$ 变大时，$-x$ 会变小，因此 $e^{-x}$ 会变小，分母 $1+e^{-x}$ 也会变小，所以整个分数会变大。因此 **Sigmoid 是单调递增函数**，也就是你之前问的：**x 越大，整体上 y 越大。** 但是它不会一直线性增大，而是越来越接近 1。
w -> 加权求和 z ->Sigmoid -> 下一层 -> Sigmoid -> ... -> 预测值 ŷ -> Loss
如果经过5个Sigmoid，且sigmoid导数最大值是0.25，通过链式法则，则为0.25的五次方
一般超过5层就会梯度消失
例如：

| x | Sigmoid(x) |
|---:|---:|
| -5 | 0.0067 |
| -2 | 0.1192 |
| 0 | 0.5000 |
| 1 | 0.7311 |
| 2 | 0.8808 |
| 5 | 0.9933 |

所以可以记成：$x<0$ 时输出小于 0.5，$x=0$ 时输出等于 0.5，$x>0$ 时输出大于 0.5。

## 4. 为什么 Sigmoid 可以表示概率？

因为 Sigmoid 的输出一定在 0～1 之间，所以在**二分类问题的输出层**，可以把它解释成属于正类的概率。例如模型计算出 $z=2$，经过 Sigmoid：

$$
\sigma(2)\approx0.881
$$

就可以理解成模型认为这个样本属于正类的概率约为 **88.1%**。比如贷款违约预测中，可以表示“这个人违约的概率约为 88.1%”。

## 5. Sigmoid 的一个重要缺点

Sigmoid 最大的问题之一是**容易出现梯度消失**。它的导数是：

$$
\sigma'(x)=\sigma(x)(1-\sigma(x))
$$

当 $x$ 非常大时，Sigmoid 输出接近 1，那么 $1-\sigma(x)$ 就接近 0，所以导数也接近 0；当 $x$ 非常小时，Sigmoid 输出接近 0，导数同样接近 0。反向传播时，如果梯度一层一层乘上这些很小的数，前面的层得到的梯度就可能越来越小，这就是梯度消失。

## 6. 最后怎么记？

你可以把 Sigmoid 理解成：

$$
加权求和 \rightarrow z \rightarrow Sigmoid \rightarrow 0\sim1
$$

也就是神经元先通过权重和偏置算出一个**分数 z**，然后 Sigmoid 把这个分数压缩成 **0～1 之间的数**。所以最核心的一句话就是：**Sigmoid 是一个 S 型的非线性函数，x 越大，输出越接近 1；x 越小，输出越接近 0；x=0 时输出 0.5。**
