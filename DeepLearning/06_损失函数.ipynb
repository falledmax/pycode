# 多分类交叉熵损失介绍

* 公式中用y是真实概率（非0即1），0将错误的概率消除，只剩真实值，所以只剩正确类别

## 代码

```python
#导包
import torch
import torch.nn as nn
from numexpr.necompiler import double
```

```python
#1.手动创建样本真实值 -> 上述公式中的y
#y_true=torch.tensor([[0,1,0],[1,0,0]],dtype=torch.float)
y_true=torch.tensor([1,2])

#2.手动创建样本的预测值 ->上述公式的f(x)
y_pred=torch.tensor([[0.1,0.8,0.1],[0.7,0.2,0.1]],requires_grad=True,dtype=float)

#3.创建多分类交叉熵损失函数
criterion = nn.CrossEntropyLoss()

#4.计算损失值
loss=criterion(y_pred,y_true)
print(f'损失值{loss}')
```

# 二分类任务损失函数

* 细节：由于公式中没有包含sigmoid激活函数，所以使用BCELoss的时候，还需要手动指定sigmoid

## 代码

```python
import torch
import torch.nn as nn

# 1.设置真实值
y_true=torch.tensor([0,1,0],dtype=torch.float32)

# 2.设置预测概率值
y_pred=torch.tensor([0.69,0.53,0.71]) #为的0的概率是0.69，1的概率是0.53

# 3.创建二分类任务损失函数
criterion=nn.BCELoss()

# 4.计算损失值
loss=criterion(y_pred,y_true)
print(f'损失值：{loss}')
```
