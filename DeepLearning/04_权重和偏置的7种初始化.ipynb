# 前五种

* 随机初始化（均匀和正太）
    * 优点：能打破对称性
    * 缺点：随机范围不当可能会导致梯度问题
    * 适用场景：浅层网络或低复杂程度模型。隐藏层1-3层，总层数不超过5层
* 全1初始化
    * 优点：实现简单
    * 缺点：无法打破对称性，所有神经元更新方向相同，无法有效训练，会导致激活值在网络中呈现指数级增长，容易梯度爆炸
    * 适用场景
        * 测试或者调试：比如验证神经网络是否可以正向或者反向传播
        * 特殊模型结构：在某些稀疏网络或者特定的自定义网络中可能需要手动设置部分参数为1
        * 偏置初始化：偶尔也可以将偏置设为1，但是很少用
* 全0初始化
    * 优点：实现简单
    * 缺点：无法打破对称性，所有神经元更新方向相同，无法有效训练。
    * 适用场景：几乎不使用，仅用于偏置的初始化

* 前五种中最常用的是 ‘全0初始化’，后两种比较重要

# 后两种

* kaiming一般和relu配合比较好，其他效果一般
    * 适用relu以及relu的变体都可以，适合深层的神经网络

* xavier初始化，适用于sigmoid、tanh等激活函数，解决梯度消失
    * 对relu等激活函数表现欠佳
    * 适用于深层网络（10层以及10以上），使用sigmoid或tanh激活函数

*  fan_in：当前层权重的输入神经元总数（上一层输出维度个数）
* fan_out：当前层权重的输出神经元总数（本层输出维度，是下一层的 fan_in）
